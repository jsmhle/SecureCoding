```python
import cv2
import numpy as np
from mtcnn import MTCNN
from tensorflow.keras.models import load_model

# 학습한 모델 불러오기
model = load_model('xception_deepfake.h5')  # Colab에 먼저 업로드해야 함

detector = MTCNN()

def predict_image_with_crop(img_path):
    img = cv2.imread(img_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    result = detector.detect_faces(img_rgb)
    if not result:
        return "❌ 얼굴을 감지하지 못했습니다."

    x, y, w, h = result[0]['box']
    face = img_rgb[y:y+h, x:x+w]
    face = cv2.resize(face, (299, 299))
    face = face / 255.0
    face = np.expand_dims(face, axis=0)

    pred = model.predict(face)[0][0]
    label = "🟢 진짜" if pred < 0.5 else "🔴 딥페이크"
    return f"{label} (신뢰도: {pred:.4f})"
```
